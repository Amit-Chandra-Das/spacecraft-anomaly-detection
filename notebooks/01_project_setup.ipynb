```python
import pandas as pd

data = {
    "Temperature": [21.4, 21.5, 21.6, 22.1, 22.3],
    "Voltage": [4.2, 4.1, 4.3, 4.2, 4.4]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df["Temperature"])

print(df["Voltage"])
```

```python
print("Temperature mean:", df["Temperature"].mean())
print("Temperature maximum:", df["Temperature"].max())
print("Temperature minimum:", df["Temperature"].min())
```

```python
print("Voltage mean:", df["Voltage"].mean())
print("Voltage maximum:", df["Voltage"].max())
print("Voltage minimum:", df["Voltage"].min())
```

```python
print(df.iloc[0])
```

```python
print(df.iloc[0:3])

print(df.iloc[2])
```

```python
print(df.head())

```

```python
print(df.shape)
```

```python
df.info()
```

```python
print("Shape:", df.shape)
print()
print("First 3 rows:")
print(df.head(3))
print()
print("Temperature average:", df["Temperature"].mean())
print("Voltage average:", df["Voltage"].mean())
print()
print("Statistics:")
print(df.describe())
```
