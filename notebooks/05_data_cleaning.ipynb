```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

project_path = '/content/drive/MyDrive/Spacecraft-Anomaly-Detection'

data_path = project_path + '/data/raw/archive/data/data'
train_path = data_path + '/train'
test_path = data_path + '/test'

print("Project loaded successfully.")
print("Train path:", train_path)
print("Test path:", test_path)
```

```python
# Load A-8 test telemetry
test_A8 = np.load(test_path + '/A-8.npy')

# Use the first column as the telemetry signal
telemetry_A8 = test_A8[:, 0]

print("Shape:", telemetry_A8.shape)
print("Minimum:", np.min(telemetry_A8))
print("Maximum:", np.max(telemetry_A8))
print("Mean:", np.mean(telemetry_A8))
print("Standard deviation:", np.std(telemetry_A8))
```

```python
# Check for missing values
missing_values = np.isnan(telemetry_A8).sum()

print("Missing values:", missing_values)
```

```python
# Check for infinite values
infinite_values = np.isinf(telemetry_A8).sum()

print("Infinite values:", infinite_values)
```

```python
# Check consecutive duplicate values
duplicate_points = np.sum(telemetry_A8[1:] == telemetry_A8[:-1])

print("Consecutive duplicate points:", duplicate_points)
```

```python
# Check the data type
print("Data type:", telemetry_A8.dtype)

# Check whether all values are numeric
print("All values are numeric:", np.issubdtype(telemetry_A8.dtype, np.number))
```

```python
# Check values outside the expected range
outside_range = np.sum(
    (telemetry_A8 < -1) | (telemetry_A8 > 1)
)

print("Values outside [-1, 1]:", outside_range)
```

```python
# Find values outside the expected range
outside_values = telemetry_A8[
    (telemetry_A8 < -1) | (telemetry_A8 > 1)
]

print("Outside values:")
print(outside_values)
```

```python
cleaning_summary = {
    "Channel": "A-8",
    "Total points": len(telemetry_A8),
    "Missing values": np.isnan(telemetry_A8).sum(),
    "Infinite values": np.isinf(telemetry_A8).sum(),
    "Consecutive duplicates": np.sum(telemetry_A8[1:] == telemetry_A8[:-1]),
    "Values outside expected range": outside_range
}

cleaning_summary
```

```python
# Final data-quality check
data_is_clean = (
    not np.isnan(telemetry_A8).any()
    and not np.isinf(telemetry_A8).any()
)

print("A-8 data ready for preprocessing:", data_is_clean)
```
