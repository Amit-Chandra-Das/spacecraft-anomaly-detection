```python
from google.colab import drive
drive.mount('/content/drive')

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

project_path = '/content/drive/MyDrive/Spacecraft-Anomaly-Detection'

data_path = project_path + '/data/raw/archive/data/data'
train_path = data_path + '/train'
test_path = data_path + '/test'

print("Project loaded successfully.")
print("Train path:", train_path)
print("Test path:", test_path)
```

```python
# List all training telemetry files

train_files = [
    file for file in os.listdir(train_path)
    if file.endswith('.npy')
]

print("Number of training channels:", len(train_files))
print("\nFirst 10 channels:")

for file in train_files[:10]:
    print(file)
```

```python
np.load(test_path + '/A-8.npy')
```

```python
print("Training folder:")
print(train_path)

print("\nFiles in training folder:")
print(os.listdir(train_path)[:10])

print("\nTotal files:")
print(len(os.listdir(train_path)))
```

```python
# Load the first training channel
first_file = train_files[0]

channel_data = np.load(train_path + '/' + first_file)

print("File:", first_file)
print("Data type:", type(channel_data))
print("Shape:", channel_data.shape)
```

```python
print(channel_data[:5])
```

```python
# Extract the main telemetry signal
telemetry_E13 = channel_data[:, 0]

print("Telemetry shape:", telemetry_E13.shape)
print("First 10 telemetry values:")
print(telemetry_E13[:10])
```

```python
# Load the first 5 training channels automatically

loaded_channels = {}

for file in train_files[:5]:
    channel_name = file.replace('.npy', '')
    data = np.load(train_path + '/' + file)

    loaded_channels[channel_name] = data

print("Loaded channels:", list(loaded_channels.keys()))

for name, data in loaded_channels.items():
    print(name, "->", data.shape)
```

```python
# Extract the main telemetry signal from each channel

telemetry_channels = {}

for name, data in loaded_channels.items():
    telemetry_channels[name] = data[:, 0]

for name, signal in telemetry_channels.items():
    print(name, "->", signal.shape)
```

```python
# Create a summary for the loaded telemetry channels

for name, signal in telemetry_channels.items():
    print(
        f"{name}: "
        f"min={np.min(signal):.3f}, "
        f"max={np.max(signal):.3f}, "
        f"mean={np.mean(signal):.3f}, "
        f"std={np.std(signal):.3f}"
    )
```

```python
constant_channels = []

for file in train_files:
    data = np.load(train_path + '/' + file)
    signal = data[:, 0]

    if np.std(signal) == 0:
        constant_channels.append(file.replace('.npy', ''))

print("Constant channels:", len(constant_channels))
print(constant_channels)
```

```python
total_channels = len(train_files)
constant_count = len(constant_channels)

constant_percentage = (constant_count / total_channels) * 100

print("Total channels:", total_channels)
print("Constant channels:", constant_count)
print(f"Constant percentage: {constant_percentage:.2f}%")
```

```python
variable_channels = [
    file for file in train_files
    if file.replace('.npy', '') not in constant_channels
]

print("Variable channels:", len(variable_channels))
print(variable_channels)
```

```python
channel_summary = []

for file in train_files:
    data = np.load(train_path + '/' + file)
    signal = data[:, 0]

    channel_summary.append({
        "Channel": file.replace(".npy", ""),
        "Time Steps": len(signal),
        "Features": data.shape[1],
        "Status": "Constant" if np.std(signal) == 0 else "Variable"
    })

channel_summary_df = pd.DataFrame(channel_summary)

print(channel_summary_df.head(10))
```

```python
# Save channel summary

summary_path = project_path + '/results/channel_summary.csv'

channel_summary_df.to_csv(summary_path, index=False)

print("Channel summary saved to:")
print(summary_path)
```
